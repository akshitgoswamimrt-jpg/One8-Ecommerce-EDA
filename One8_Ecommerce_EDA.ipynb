# Data‑Driven Insights into One8’s E‑Commerce Performance
--------------------------------------------------------------------

# Project Overview

This project is an end-to-end Exploratory Data Analysis (EDA) of the **One8** e-commerce website, a fashion and sportswear brand. The objective was to collect product data directly from the website using web scraping, clean and prepare the dataset, perform exploratory data analysis, visualize important trends, and generate business insights that can support data-driven decision making.

Instead of working with a ready-made dataset, I built my own dataset by scraping product information from the live website. This project demonstrates the complete workflow of a Data Analyst, from data collection to business insight generation.

---

## Problem Statement

E-commerce companies manage hundreds of products across multiple categories, making it difficult to manually understand pricing strategies, product availability, category performance, and product distribution.

The goal of this project was to answer important business questions using data analysis.

---

## Business Questions Solved

* How many products are available on the website?
* Which product categories contain the most products?
* What is the average price of products in each category?
* Which category is the most expensive and which is the least expensive?
* What are the minimum, maximum, mean, and median product prices?
* How are products distributed across different price ranges?
* Which products have the highest number of variants?
* Which products contain the highest number of images?
* Are there duplicate or missing records in the dataset?
* What business insights can be extracted from the collected data?

# Data Visualization :-
---------------------------
Created various visualizations using Matplotlib to represent product distribution, category-wise analysis, price distribution, and average pricing. These charts helped simplify complex data and provided a clear understanding of product trends.


```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
plt.figure(figsize=(10,5))

df["Category"].value_counts().plot(kind="bar")

plt.title("Number of Products in Each Category")
plt.xlabel("Category")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Price"], bins=10)

plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
avg_price = df.groupby("Category")["Price"].mean().sort_values()

plt.figure(figsize=(10,5))

avg_price.plot(kind="barh")

plt.title("Average Price by Category")
plt.xlabel("Average Price")
plt.ylabel("Category")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["Price Range"].value_counts().sort_index().plot(kind="bar")

plt.title("Products by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Count")

plt.show()
```

```python
top10 = df.sort_values("Price", ascending=False).head(10)

plt.figure(figsize=(10,6))

plt.barh(top10["Product Name"], top10["Price"])

plt.title("Top 10 Most Expensive Products")
plt.xlabel("Price")
plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.show()
```
